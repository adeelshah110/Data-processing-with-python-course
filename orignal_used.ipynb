## Part 1: Reading the Dataset

In this module, we will work with the Multi-Hospital
Lab Results dataset.

The dataset contains laboratory records collected
from three hospitals.

We will use Pandas to read and explore the dataset
before performing any data cleaning.

**Dataset source:** Kaggle - Multi-Hospital Lab Results
(Messy Data)

https://www.kaggle.com/datasets/nudratabbas/multi-hospital-lab-results-messy-data

```python
#example Dataset one 
import pandas as pd

# Read the dataset
df = pd.read_csv("Dataset/Hospital_lab/multi_hospital_lab_results.csv")

# Display first five rows
display(df.head())

# Part 2 understanding the structure of dataset
# 







```

```python
#Number of rows and columns
#print("Dataset shape:", df.shape)

df.columns

# Column names
print("\nColumn names:")
print(df.columns.tolist())

# Data types
print("\nData types:")
print(df.dtypes)

```

```python
# General information
df.info()
```

```python


# Part 3 
missing_values = df.isnull().sum()

print(missing_values)
```

```python

#checking for duplicates
duplicates = df.duplicated().sum()

print(duplicates)
```

```python


# Part 4 unique and non unique counts
# Unique hospitals
print(df["hospital"].unique())

# Number of hospitals
print(df["hospital"].nunique())

# Number of records per hospital
print(df["hospital"].value_counts())

#
```

```python
# Unique test names
print(df["test_name"].unique())

# Number of unique test names
print(df["test_name"].nunique())

# Frequency of each test name
print(df["test_name"].value_counts())
# Unique measurement units
print(df["unit"].unique())

# 
```

```python
#Frequency of units
print(df["unit"].value_counts())

# Display test names, values, and units
display(df[["test_name", "test_value", "unit"]].head(10))

```

```python
# Part 5

# Number of unique patient IDs
print("Unique patient IDs:", df["patient_id"].nunique())

# Total laboratory records
```

```python
print("Total records:", len(df))

# First ten collection dates
print(df["collection_date"].head(10))
```

```python


# part 6 Describe command
# Numerical summary
display(df.describe())
```
