# 2. Mathematical Operations

Python supports basic mathematical operations.

These operations are useful for calculations and data analysis.

```python
a = 20
b = 6

print(a + b) # Addition
print(a - b) # Subtraction
print(a * b) # Multiplication
print(a / b) # Division
print(a // b) # Floor division
print(a % b) # Remainder
print(a ** b) # Power
```

```python
# Example 2
price = 100
quantity = 5

total = price * quantity

print("Total Price:", total)
```

```python

# Example 3
value1 = 10
value2 = 20
value3 = 30

average = (value1 + value2 + value3) / 3

print("Average:", average)
```

## 3. Type Conversion

Type conversion means changing data from one type to another.

Common functions:
- int() converts to integer
- float() converts to decimal number
- str() converts to string

Type conversion is important when cleaning datasets.

```python
age = "35"

print(type(age))
age = int(age)

print(age)
print(type(age))
```

```python
price = "125.50"

price = float(price)

print(price)
print(type(price))
```

```python


patient_id = 1001

patient_id = str(patient_id)

print(patient_id)
print(type(patient_id))
```

```python


value = "12.5"

# int(value) will produce an error

value = float(value)

print(value)
```

## 4. Working with Strings

Strings are used to store text data.

Python provides functions to clean and transform text.

Common operations:
- upper()
- lower()
- title()
- strip()
- replace()
- split()

```python
hospital = "Hospital A"

print(hospital.upper())
print(hospital.lower())
print(hospital.title())


```

```python
hospital = " Hospital A "

print(hospital.strip())
hospital = "Hospital_A"

print(hospital.replace("_", " "))
```

```python


record = "Hospital_A,101,Glucose,5.6"
print(type(record))
values = record.split(",")

print(values)
print(type(values))
```

## 5. Lists

A list stores multiple values in one variable.

Lists are useful for storing collections of data.

We can:
- Access elements

- Remove elements
- Update values
- Sort values

```python
# Section 5 List 
# Example 1
hospitals = ["Hospital_A", "Hospital_B", "Hospital_C"]
print(hospitals)

print(type(hospitals))
# hostpital positions
print(hospitals[0])
print(hospitals[1])
print(hospitals[-1])
# Add to


```

## 6. Dictionaries

A dictionary stores data using key-value pairs.

Dictionaries are useful for representing structured records.

For example, one hospital laboratory record can contain:
- Hospital name
- Patient ID
- Test name
- Test value

```python
patient = {
    "hospital": "Hospital_A",
    "patient_id": 101,
    "test_name": "Glucose",
    "test_value": 5.6
}

print(patient)
```

```python
print(type(patient))
```

```python


print(patient["hospital"])
print(patient["patient_id"])
print(patient["test_value"])

```

```python
patient["test_value"] = 6.2

print(patient)
patient["unit"] = "mmol/L"

print(patient)


```

```python
#print(patient.keys())
#print(patient.values())
```

```python
print(patient.items())
```

## 7. Combining Lists and Dictionaries

Real-world datasets contain multiple records.

We can store multiple dictionaries inside a list.

Each dictionary represents one record.

```python
patients = [
    {"patient_id": 101, "hospital": "Hospital_A", "test_value": 5.6},
    {"patient_id": 102, "hospital": "Hospital_B", "test_value": 7.2},
    {"patient_id": 103, "hospital": "Hospital_C", "test_value": 4.8}
]

print(patients)
print(patients[0])
```

```python

print(patients[1])
print(patients[0]["hospital"])
print(patients[1]["test_value"])
```
