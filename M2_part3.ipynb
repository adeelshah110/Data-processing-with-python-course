# Part 1: Introduction to TXT Files and Text Encoding

Text files (`.txt`) are commonly used to store logs, reports, scientific measurements, and other data.

Python can read TXT files using the built-in `open()` function or Pandas.





### Advantages and disadvantages of TXT files

| **Advantages** | **Disadvantages** |
|:---|:---|
| **Simple format:** Easy to create, read, and edit. | **No fixed structure:** Difficult to organize complex data. |
| **Lightweight:** Usually has a small file size. | **No data types:** Numbers and dates are initially read as text. |
| **Widely supported:** Works on almost every operating system. | **Manual processing:** May require splitting and cleaning text. |
| **Easy to process:** Python provides built-in functions for reading TXT files. | **Encoding issues:** Special characters may cause errors. |
| **Human-readable:** Can be opened using Notepad or any text editor. | **Limited features:** No built-in tables, formulas, or metadata structure. |
| **Useful for raw data:** Suitable for logs, notes, and text analysis. | **Inconsistent formatting:** Different lines may have different structures. |


## What is Text Encoding?

Text encoding determines how characters are stored as bytes.

| Encoding | Description |
|---|---|
| ASCII | Basic English characters |
| UTF-8 | Supports multilingual characters |
| ISO-8859-1 | Older Western European encoding |
| Windows-1252 | Common in older Windows files |
| UTF-16 | Unicode encoding |



**Example:** Café, Hyvää päivää, José

Using an incorrect encoding can produce unreadable characters or a `UnicodeDecodeError`.

UTF-8 is generally recommended for modern data processing.


```python

# Create a UTF-8 file
with open("Dataset/students_utf8.txt", "w", encoding="utf-8") as f:
    f.write("Name: Ahmad\n")
    f.write("City: Kuopio\n")
    f.write("Text: Hyvää päivää\n")

# Create an ISO-8859-1 file
with open("Dataset/students_latin1.txt", "w", encoding="iso-8859-1") as f:
    f.write("Name: André\n")
    f.write("City: Montréal\n")
    f.write("Text: Café\n")

# Create a UTF-16 file
with open("Dataset/students_utf16.txt", "w", encoding="utf-16") as f:
    f.write("Name: Aisha\n")
    f.write("City: Helsinki\n")
    f.write("Text: Hello World\n")
```


# Part 2: Reading TXT Files Using Python

Python provides the `open()` function for reading and writing text files.

Common file modes:

- `"r"` — Read a file
- `"w"` — Write a file
- `"a"` — Append data to a file

Two common reading methods:

- `read()` — Read the entire file
- `for line in file` — Read line by line

The `with` statement automatically closes the file.


```python

with open("Dataset/students_utf8.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

```python

with open("Dataset/students_utf8.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line)
        #print(line.strip())
```

```python
# test this with 8 first
with open("Dataset/students_utf16.txt", "r", encoding="utf-16") as file:
    print(file.read())
```


# Part 3: Encoding Errors and Their Solutions

A `UnicodeDecodeError` occurs when Python cannot decode file bytes using the selected encoding.

For example, a file saved using ISO-8859-1 may produce an error when read using UTF-8.

## Common Solutions

| Method | Description |
|---|---|
| Correct encoding | Reads the original characters correctly |
| `errors="ignore"` | Skips invalid characters |
| `errors="replace"` | Replaces invalid characters with � |

**Important:** Ignoring or replacing characters may result in information loss.


```python

# This code produces UnicodeDecodeError
with open("Dataset/students_latin1.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

```python

with open("Dataset/students_latin1.txt", "r", encoding="iso-8859-1") as file:
    print(file.read())
```

```python

with open("Dataset/students_latin1.txt", "r",
          encoding="utf-8", errors="ignore") as file:
    print(file.read())
```

```python

with open("Dataset/students_latin1.txt", "r",
          encoding="utf-8", errors="replace") as file:
    print(file.read())
```


# Part 4: Detecting and Converting Text Encoding

Sometimes, the encoding of a text file is unknown.

The `charset-normalizer` library can estimate file encoding.

We can also convert older text files into UTF-8 for better compatibility.

**Note:** Automatic encoding detection is not always accurate. Always verify the decoded text.


```python

%pip install charset-normalizer

from charset_normalizer import from_path

result = from_path("Dataset/students_latin1.txt").best()

if result is not None:
    print("Detected encoding:", result.encoding)
    print("Text:", str(result))
else:
    print("Encoding could not be detected")


```

```python

# Read using original encoding
with open("Dataset/students_latin1.txt", "r", encoding="iso-8859-1") as f:
    text = f.read()

# Save using UTF-8
with open("Dataset/students_converted.txt", "w", encoding="utf-8") as f:
    f.write(text)

print("File converted successfully")
```

```python

with open("Dataset/students_converted.txt", "r", encoding="utf-8") as f:
    print(f.read())
```


# Part 5: Reading TXT Files Using Pandas

TXT files may contain structured data organized into rows and columns.

Common separators:

- Comma `,`
- Semicolon `;`
- Tab `\t`
- Pipe `|`

Pandas provides `pd.read_csv()` to import structured TXT files.

The `sep` parameter specifies the column separator.


```python

with open("Dataset/students.txt", "w", encoding="utf-8") as f:
    f.write("Name;Age;City\n")
    f.write("Ahmad;25;Kuopio\n")
    f.write("Anna;22;Helsinki\n")
    f.write("Maria;24;Joensuu\n")
```

```python

import pandas as pd

df = pd.read_csv("Dataset/students.txt", sep=";", encoding="utf-8")

df.head()
```

```python

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("Data types:")
print(df.dtypes)
```

```python

df.isna().sum()
```


# Part 6: Handling File Errors and Reusable Functions

File reading may fail because:

- The file does not exist.
- The selected encoding is incorrect.

Python uses `try-except` to handle such errors.

We can also create reusable functions to read multiple text files without repeating code.


```python

try:
    with open("students_utf8.txt", "r", encoding="utf-8") as file:
        print(file.read())

except FileNotFoundError:
    print("File not found")

except UnicodeDecodeError:
    print("Incorrect file encoding")
```

```python

def read_text_file(filepath, encoding="utf-8"):
    try:
        with open(filepath, "r", encoding=encoding) as file:
            return file.read()
    except (FileNotFoundError, UnicodeDecodeError) as error:
        print(error)
        return None

content = read_text_file("Dataset/students_utf8.txt")
print(content)
```
