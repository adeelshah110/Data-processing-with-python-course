## Part 1: Introduction to Regular Expressions

### What Are Regular Expressions?

A **Regular Expression (Regex)** is a sequence of characters that defines a search pattern.

Regular expressions are used to find, identify, extract, and sometimes replace specific patterns in text.

Instead of searching for an exact word, regex allows us to search for patterns such as:

- Email addresses
- Phone numbers
- Dates
- Numbers
- Words starting with a specific letter
- Text containing specific characters

### Why Do We Need Regular Expressions?

In real-world data processing, text data often contains information in different formats.

For example, a dataset may contain:
- Different phone number formats
- Email addresses inside longer text
- Dates written in different ways
- Unwanted characters
- Text containing numbers and symbols

Regular expressions help us identify these patterns efficiently.

### Regular Expressions in Python

Python provides the built-in `re` module for working with regular expressions.

We can use this module to:
- Search for patterns
- Find all matching values
- Extract information from text
- Replace matching patterns
- Validate text formats

### Important Note

Regular expressions work with patterns, not just exact text.



For example:
- `cat` searches for the exact sequence of characters "cat".
- `\d` matches a digit.
- `\d+` matches one or more digits.
- `\w+` matches one or more word characters.


```python

import re

# Example 1: Searching for a word
text = "Python is useful for data processing."
result = re.search("Python", text)
print("Example 1:", result)
```

```python


# Example 2: Extracting numbers
text = "The hospital recorded 25 patients in room 3 and 120 patients in room 5."
numbers = re.findall(r"\d+", text)
print("Example 2:", numbers)
print (type(numbers[0]))
#
```

```python
# Convert extracted numbers into integers
numbers = [int(number) for number in numbers]
print("Numbers as integers:", numbers)
print (type(numbers[0]))
#
```

```python
# Example 3: Extracting words
text = "Data processing with Python is interesting!"
words = re.findall(r"\w+", text)
print("Example 3:", words)
```

```python

# Example 4: Case-insensitive search
text = "Python is popular. python is easy. PYTHON is powerful."
result = re.findall(r"python", text, re.IGNORECASE)
print("Example 4:", result)

#
```

```python
#Example 5: Replacing numbers
text = "The hospital has 25 patients and 3 doctors."
cleaned_text = re.sub(r"\d+", "NUMBER", text)
print("Example 5:", cleaned_text)

```


## Extracting Email Addresses Using Regular Expressions

An email address generally contains three main parts:

1. Username
2. @ symbol
3. Domain name and extension

Example:

`john.smith@gmail.com`

We can use regular expressions to identify and extract email addresses from text.

### Important Regex Patterns

| Pattern | Meaning |
|---|---|
| `\w` | Matches letters, digits, and underscore |
| `+` | Matches one or more occurrences |
| `.` | Matches any character except newline |
| `\.` | Matches an actual dot |
| `@` | Matches the @ symbol |
| `[A-Za-z]` | Matches uppercase or lowercase letters |
| `[0-9]` | Matches digits |
| `{2,}` | Matches at least two occurrences |
| `[abc]` | Matches any one character from a, b, or c |
| `[^abc]` | Matches characters other than a, b, or c |

### Why Do We Use `r` Before a Regex?

In Python, `r` creates a raw string.

For example:

`r"\w+"`

It helps Python interpret backslashes correctly in regular expressions.


```python
text = """
Contact John at john.smith@gmail.com
Support: support@uef.fi
Research: research_team123@hospital.org
Student: ali123@gmail.com
Invalid email: john@
"""

print("Original Text:\n", text)

# Pattern 1:
```

```python
#Find @ symbols
print("\n1. Finding @ symbols:")
print(re.findall(r"@", text))

# Pattern 2: Find words
print("\n2. Finding words:")
print(re.findall(r"\w+", text))
```

```python
# Pattern 3: Find words ending with @
print("\n3. Finding text before @:")
print(re.findall(r"\w+@", text))
```

```python
# Pattern 4: Find username and domain without extension
print("\n4. Finding username and domain:")
print(re.findall(r"\w+@\w+", text))
```

```python
# Pattern 5: Find simple email addresses
print("\n5. Simple email extraction:")
print(re.findall(r"\w+@\w+\.\w+", text))

#
```

```python
# Pattern 6: Extract common email formats
print("\n6. Complete email extraction:")
pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"
emails = re.findall(pattern, text)
print(emails)
```

```python


# Pattern 7: Count extracted emails
print("\n7. Number of emails:")
print(len(emails))

# Pattern 8: Extract emails without case sensitivity
print("\n8. Case-insensitive email extraction:")
print(re.findall(pattern, text, re.IGNORECASE))

# 
```

```python
#Pattern 9: Replace email addresses with EMAIL
print("\n9. Replace emails:")
cleaned_text = re.sub(pattern, "EMAIL", text)
print(cleaned_text)
```

```python
# Pattern 10: Extract only usernames
print("\n10. Extract usernames:")
usernames = re.findall(r"([A-Za-z0-9._%+-]+)@", text)
print(usernames)

```

```python
# Pattern 11: Extract domain names
print("\n11. Extract domains:")
domains = re.findall(r"@([A-Za-z0-9.-]+\.[A-Za-z]{2,})", text)
print(domains)
#
```

```python
# Pattern 12: Extract unique email addresses
print("\n12. Unique emails:")
unique_emails = list(set(emails))
print(unique_emails)
```
