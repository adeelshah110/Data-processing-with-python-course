## 1. Your First Python Program

### The `print()` Function

The `print()` function displays text, numbers, or calculation results.

Text values must be enclosed in quotation marks, either single quotes (`' '`) or double quotes (`" "`).

Python executes instructions from top to bottom within a code cell.


````python
#part 1 print function
print("Hello, Python!")
print("Welcome to Data Processing with Python")
print("University of Eastern Finland")

print(10)
print(10 + 20)
display("text")
````

````python
student_name = "Anna"
student_age = 22
student_score = 85

print("Student:", student_name)
print("Age:", student_age)
print("Score:", student_score)

score = 70

print("Original score:", score)

score = 85

print("Updated score:", score)

score = score + 5

print("Final score:", score)
````


## 2. Comments in Python

Comments are notes written inside a program to explain the code.

Python ignores comments during execution.

A single-line comment starts with the `#` symbol.

Comments are useful for:

- Explaining what a program does.
- Describing calculations.
- Making code easier to understand.
- Helping other people read and maintain the program.

Use short and meaningful comments rather than explaining every individual line.



## 3. Variables in Python

A variable is a name used to store a value in a program.

For example, we can use variables to store a student's name, age, or examination score.

### Creating Variables

Python uses the assignment operator `=` to assign a value to a variable.

```python
name = "Anna"
age = 22
```

In Python, we do not need to declare the data type of a variable before assigning a value.

### Rules for Variable Names

- A variable name can contain letters, numbers, and underscores.
- A variable name cannot start with a number.
- Variable names cannot contain spaces.
- Python variable names are case-sensitive.
- Python keywords such as `if`, `for`, and `class` cannot be used as variable names.

### Recommended Naming Style

Use meaningful variable names with underscores.

Examples:

- `student_name`
- `patient_age`
- `sample_temperature`
- `total_score`


## 4. Basic Data Types

Every value in Python has a data type.

The data type determines what kind of information a variable stores and which operations can be performed on it.

### Common Python Data Types

| Data Type | Description | Example |
|---|---|---|
| `int` | Integer or whole number | `25` |
| `float` | Decimal number | `36.5` |
| `str` | Text or string | `"Anna"` |
| `bool` | Boolean value | `True` or `False` |
| `list` | Ordered collection | `[10, 20, 30]` |
| `dict` | Key-value collection | `{"name": "Anna"}` |
| `NoneType` | Represents no value | `None` |

### The `type()` Function

The `type()` function identifies the data type of a value.

Understanding data types is important because data-processing operations often require particular types.

For example, we cannot directly add a string containing `"20"` to the integer `10`.
