
# Module 2: Reading and Understanding Datasets

## Part 1: Reading the Dataset

In this module, we will work with the Multi-Hospital
Lab Results dataset.

The dataset contains laboratory records collected
from three hospitals.

We will use Pandas to read and explore the dataset
before performing any data cleaning.

**Dataset source:** Kaggle - Multi-Hospital Lab Results
(Messy Data)

https://www.kaggle.com/datasets/nudratabbas/multi-hospital-lab-results-messy-data


```python
# Example Dataset one 
import pandas as pd

# Read the dataset
df = pd.read_csv("Dataset/Hospital_lab/multi_hospital_lab_results.csv")

# Display first five rows
display(df.head())
```


## Understanding Rows and Columns

A DataFrame organizes information into rows and columns.

- **Rows:** Individual laboratory records.
- **Columns:** Attributes describing each record.
- **Index:** Row labels used by Pandas.

Each row represents one laboratory test record.
A patient may appear in multiple records.



# Part 2: Understanding Dataset Structure

Before processing a dataset, we need to understand
its size, columns, and data types.

We will use:

- `shape` - Number of rows and columns
- `columns` - Column names
- `dtypes` - Data types
- `info()` - General dataset information


```python

# Number of rows and columns
print("Dataset shape:", df.shape)

# Column names
print("\nColumn names:")
print(df.columns.tolist())

# Data types
print("\nData types:")
print(df.dtypes)

# General information
df.info()
```


## Understanding Data Types

Pandas identifies the data type of each column.

In our dataset:

- `patient_id` is stored as an integer.
- `test_value` is stored as a string.
- `collection_date` is stored as a string.
- Other descriptive columns contain text.

Data types are important because they determine
which operations we can perform on the data.

We will investigate incorrect data types in Module 3.



## 3.1 Displaying Dataset Records

We use:

- `head()` to display the first rows.
- `tail()` to display the last rows.
- `sample()` to display randomly selected rows.

These methods help us inspect the dataset
without modifying its contents.



## 3.2 Exploring Hospital Records

We will examine how laboratory records are
distributed across hospitals.

We use:

- `unique()` to identify hospital names.
- `nunique()` to count different hospitals.
- `value_counts()` to count records per hospital.




```python
missing_values = df.isnull().sum()

print(missing_values)

#checking for duplicates
duplicates = df.duplicated().sum()

print(duplicates)
```


## Exploring Dataset Columns

We will explore the hospital lab dataset to understand
the values stored in different columns.

We will use:
- `unique()` to display unique values.
- `nunique()` to count unique values.
- `value_counts()` to count repeated values.


```python

# Unique hospitals
print(df["hospital"].unique())

# Number of hospitals
print(df["hospital"].nunique())

# Number of records per hospital
print(df["hospital"].value_counts())
```


## 3.3 Exploring Laboratory Test Names

The `test_name` column describes the laboratory
test associated with each record.

We will identify:

- Different test names.
- Number of unique test names.
- Frequency of each test name.

This helps us understand the types of laboratory
measurements represented in the dataset.


```python

# Unique test names
print(df["test_name"].unique())

# Number of unique test names
print(df["test_name"].nunique())

# Frequency of each test name
print(df["test_name"].value_counts())
```


## 3.4 Exploring Measurement Units

Laboratory measurements may use different units.

The `unit` column stores the measurement unit
associated with each laboratory result.

We will inspect the available units and
their frequencies.


```python

# Unique measurement units
print(df["unit"].unique())

# Frequency of units
print(df["unit"].value_counts())

# Display test names, values, and units
display(df[["test_name", "test_value", "unit"]].head(10))
```


## 3.5 Exploring Patients and Collection Dates

The dataset includes patient identifiers and
sample collection dates.

We will check:

- Number of unique patient identifiers.
- Number of laboratory records.
- Examples of collection dates.

Patient identifiers are used to distinguish
patients within the dataset.


```python

# Number of unique patient IDs
print("Unique patient IDs:", df["patient_id"].nunique())

# Total laboratory records
print("Total records:", len(df))

# First ten collection dates
print(df["collection_date"].head(10))
```


## 3.6 Descriptive Summary

The `describe()` method summarizes dataset columns.

For numerical columns, it provides statistics
such as minimum, maximum, and average.

For text columns, it provides information
such as unique values and the most frequent value.

We will use it to understand the dataset
before data cleaning.


```python

# Numerical summary
display(df.describe())

# Summary of all columns
display(df.describe(include="all").T)
```


# Part 4: Initial Data Quality Observations

Before cleaning a dataset, we should inspect
its completeness and consistency.

We will check:

- Missing values
- Duplicate rows
- Column data types

Our goal is to observe possible issues,
not to correct them yet.


```python

# Missing values
print("Missing values:")
print(df.isna().sum())

# Duplicate rows
print("\nDuplicate rows:")
print(df.duplicated().sum())

# Data types
print("\nData types:")
print(df.dtypes)
```
